# Python Lists

### Creating Python Lists

A list in Python is created by placing all the items (elements) inside square brackets `[]`, separated by commas.

**Key features:**
* Lists can contain items of different data types (integers, strings, floats, even other lists).
* Lists are ordered and mutable (modifiable).

**Syntax:**
```python
my_list = [item1, item2, item3]
```

````python
# Empty list
empty_list = []

# List of integers
numbers = [1, 2, 3, 4, 5]

# List of strings
fruits = ["apple", "banana", "cherry"]

# Mixed data types
mixed_list = [10, "hello", 3.14, True]

# Displaying the lists
print("Numbers:", numbers)
print("Fruits:", fruits)
print("Mixed:", mixed_list)
````
