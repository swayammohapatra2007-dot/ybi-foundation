# Pandas Data Summary

```python
import pandas as pd

# Create a sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [25, 30, 35, 40, 22],
    'Salary': [50000, 60000, 80000, 120000, 45000],
    'Department': ['HR', 'Engineering', 'Engineering', 'Management', 'HR']
}
df = pd.DataFrame(data)

# 1. Display the first few rows
print("--- First 3 Rows ---")
display(df.head(3))

# 2. Display DataFrame info (columns, data types, non-null counts)
print("\n--- DataFrame Info ---")
df.info()

# 3. Descriptive statistics for numerical columns
print("\n--- Statistical Summary ---")
display(df.describe())

# 4. Summary of categorical columns
print("\n--- Categorical Summary ---")
display(df.describe(include=['O']))
```

```python
# 5. Value counts for categorical columns
print("--- Department Value Counts ---")
display(df['Department'].value_value_counts() if hasattr(df['Department'], 'value_value_counts') else df['Department'].value_counts())

# 6. Groupby summaries (e.g., average age and salary by department)
print("\n--- Groupby Summary (Department) ---")
display(df.groupby('Department')[['Age', 'Salary']].mean())

# 7. Check for missing values
print("\n--- Missing Values Count ---")
display(df.isnull().sum())

# 8. Numeric correlation matrix
print("\n--- Correlation Matrix ---")
display(df.corr(numeric_only=True))
```
