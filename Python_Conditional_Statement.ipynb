# Python Conditional Statement

### Understanding Python `if-else` Statements

Conditional statements allow your program to make decisions and execute different blocks of code based on whether a condition is **True** or **False**.

#### 1. The Basic `if` Statement
If the condition evaluates to `True`, the indented block of code directly below it is executed.

#### 2. The `if-else` Statement
Provides an alternative block of code that runs if the initial condition is `False`.

#### 3. The `if-elif-else` Chain
Used to check multiple conditions sequentially. `elif` stands for "else if". The first condition that evaluates to `True` gets executed, and the rest are ignored.

```python
# Example of an if-else statement
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: F")
```
