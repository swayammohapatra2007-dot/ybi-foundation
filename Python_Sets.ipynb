# Python Sets

```python
# Python Sets Examples

# 1. Creating a Set
# Sets are unordered, unchangeable (but you can remove and add items), and do not allow duplicate values.
my_set = {1, 2, 3, 3, 4}  # Duplicate 3 will be ignored
print("Initial Set:", my_set)

# 2. Adding items
my_set.add(5)
print("After adding 5:", my_set)

# 3. Removing items
my_set.discard(2)  # Removes 2. discard() won't raise an error if item is missing.
print("After removing 2:", my_set)

# 4. Set Operations
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

# Union (all unique elements from both sets)
print("Union:", set_a.union(set_b))

# Intersection (elements present in both sets)
print("Intersection:", set_a.intersection(set_b))

# Difference (elements in set_a but not in set_b)
print("Difference (A - B):", set_a.difference(set_b))

# Symmetric Difference (elements in either set, but not both)
print("Symmetric Difference:", set_a.symmetric_difference(set_b))
```

```python
# More Python Set Examples

# 1. Membership Testing (Very fast in sets compared to lists)
fruits = {"apple", "banana", "cherry"}
print("Is 'apple' in fruits?", "apple" in fruits)
print("Is 'grape' in fruits?", "grape" in fruits)

# 2. Subset and Superset relationships
set_x = {1, 2}
set_y = {1, 2, 3, 4}

print("Is set_x a subset of set_y?", set_x.issubset(set_y))       # True
print("Is set_y a superset of set_x?", set_y.issuperset(set_x))   # True
print("Are set_x and set_y disjoint?", set_x.isdisjoint(set_y))   # False (they share elements)

# 3. Set Comprehension
squared_set = {x**2 for x in [1, 2, 2, 3, 4]}  # Note duplicates are automatically handled
print("Squared set comprehension:", squared_set)
```
