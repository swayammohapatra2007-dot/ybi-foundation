# Python Sets

```python
# Creating a set
my_set = {1, 2, 3, 4, 5}
print("Initial Set:", my_set)

# Adding elements
my_set.add(6)
print("After adding 6:", my_set)

# Removing elements
my_set.remove(3)
print("After removing 3:", my_set)

# Set Operations
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

# Union
print("Union:", set_a.union(set_b))

# Intersection
print("Intersection:", set_a.intersection(set_b))

# Difference
print("Difference (A - B):", set_a.difference(set_b))
```
