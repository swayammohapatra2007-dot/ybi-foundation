# Python Loops

### Understanding Python `for` Loops

A `for` loop in Python is used to iterate over a sequence (such as a list, tuple, dictionary, set, or string) or other iterable objects.

#### Basic Syntax:
```python
for item in sequence:
    # code block to execute
```

* **`item`**: A variable that takes the value of the element inside the sequence on each iteration.
* **`sequence`**: The collection of items you want to loop through.

#### Common Use Case: `range()`
If you want to loop through a block of code a specific number of times, you can use Python's built-in `range()` function.

````python
# Example 1: Iterating over a list
fruits = ['apple', 'banana', 'cherry']
print('Iterating over a list:')
for fruit in fruits:
    print(f'I like {fruit}')

print('\n' + '='*30 + '\n')

# Example 2: Using the range() function (loops 5 times, from 0 to 4)
print('Using range(5):')
for i in range(5):
    print(f'Current index: {i}')
````
